# ClinVar Pathogenicity Benchmark — All Models
Zero-shot cosine distance (and MLM log-likelihood for ESM1b-MLM) on 7,991 missense variants from 2,381 unique proteins.

![Pipeline](figures/project_pipeline.png)

```python
import pandas as pd
from IPython.display import display, Image
import os

FIG_DIR = 'figures'
for f in os.listdir(FIG_DIR):
    if f.endswith('.png'):
        size = os.path.getsize(os.path.join(FIG_DIR, f))
        print(f'{f:40s} {size/1024:.0f} KB')
```

## Summary Metrics

```python
df = pd.read_csv(os.path.join(FIG_DIR, 'benchmark_summary.csv'))
display(df.style.format({
    'AUROC': '{:.3f}', 'AUPRC': '{:.3f}', 'Accuracy': '{:.3f}',
    'Precision': '{:.3f}', 'Recall': '{:.3f}', 'F1': '{:.3f}',
    'MannWhitney_p': '{:.2e}', 'Spearman_p': '{:.2e}'
}).highlight_max(subset=['AUROC', 'AUPRC', 'F1'], color='#d4edda', axis=0))
```
